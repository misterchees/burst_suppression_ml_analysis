```python
from vitaldb import find_cases, VitalFile
import os
import re
import pandas as pd
```

```python
# define track names
track_names = ['Solar8000/HR']
# 1 value per sec
interval = 1

# folder to save 
output_folder = 'C:\\Users\\jesus\\OneDrive\\Dokumente\\Jesús\\Studium\\Fächer - Bioinformatik\\Praktische Arbeit und Bachelorarbeit\\Material\\Daten\\Initial data\\vitaldb_csv_cardiovasc'  # Ensure this folder exists
os.makedirs(output_folder, exist_ok=True)
```

```python
# find all cases that have defined track name
try:
    caseids = find_cases(track_names)
    print(f"Found {len(caseids)} cases with the tracks {track_names}.")

except Exception as e:
    print(F"Error querying case IDs: {e}")
    caseids = []
```

```python
# case ids are not necessarily the same number as their index. Check single id here and set caseid
print(caseids[3496])
caseid = caseids[3496]
```

```python
# Save single case here
#load file
vf = VitalFile(caseid)

# Construct the output path for the CSV file
output_path = os.path.join(output_folder, f"{caseid}.csv")

# extract track_names data longitudinally
vf.to_csv(opath= output_path,track_names=track_names, interval=interval, return_datetime=True, return_timestamp=True)
```

```python
# Path to dir with existing CSV-files
existing_csv_dir = 'C:\\Users\\jesus\\OneDrive\\Dokumente\\Jesús\\Studium\\Fächer - Bioinformatik\\Praktische Arbeit und Bachelorarbeit\\Material\\Daten\\Initial data\\vitaldb_csvprocessed_BIS_BIS_SR_MAC'

# Regex to extract number from csv e.g. "123.csv" -> 123
pattern = re.compile(r'^(\d+)\.csv$')

# Set to save all found numbers
existing_csv_set = set()

# loop through all data in existing_csv_dir
for filename in os.listdir(existing_csv_dir):
    match = pattern.match(filename)
    if match:
        number = int(match.group(1))
        existing_csv_set.add(number)
print(existing_csv_set)
```

```python
# Start processing from the first case
start_index = 0

# Debugging: check caseids
print(f"Total cases: {len(caseids)}")
print(f"Starting from case: {start_index}")

for i in range(start_index, len(caseids)):
    caseid = caseids[i]  # Get the current case ID
    if caseid in existing_csv_set:
        try:
            #load file
            vf = VitalFile(caseid)
    
            # Construct the output path for the CSV file
            output_path = os.path.join(output_folder, f"{caseid}.csv")
    
            # extract track_names data longitudinally
            vf.to_csv(opath= output_path,track_names=track_names, interval=interval, return_datetime=True, return_timestamp=True)
    
            print(f"Processed case {i + 1}/{len(caseids)}: {caseid}")
            
            # remove case from set to see the remainder when download finishes
            existing_csv_set.discard(caseid)
            
        except Exception as e:
            print(f"Error processing case {caseid}: {e}")
```
